### **Como RAG puede ser una solución de seguridad crucial**

RAG puede ser considerado como una solución a brechas de seguridad, en lugar de ser el principal causante. Esto normalmente se suele hacer limitando el acceso de datos, asegurando que las respuestas sean más fiables y siendo completamente transparente al proporcionar fuentes de datos.

#### Limitación de datos

Se implementan sistemas de autenticación por usuario tales como los de cualquier web del navegador, haciendo una restricción de los datos en el proceso de *retrieve* del RAG, asegurándonos que la información sensible se accede únicamente por el usuario en concreto (además de sistemas de encriptación de datos convencionales).

#### Garantizar la fiabilidad de las respuestas

Por el funcionamiento de un RAG, las respuestas generadas son más precisas, reduciendo así el riesgo de generar respuestas incorrectas o imprecisas. 

Además, se tiene el control de las fuentes de datos de nuestro RAG, ya que las controlamos directamente nosotros en nuestra *vector database*, por lo que si nos preocupamos por la calidad de nuestras fuentes de datos se verá reflejado en una mejor respuesta (esto es super importante en áreas donde la precisión son esenciales, como la salud, finanzas, o aplicaciones legales).

#### Mantener la transparencia

Es posible mantener la transparencia de nuestro RAG citando las fuentes de datos en las que se basa la respuesta, aumentando la confianza con el usuario demostrando la fiabilidad del contenido generado, de manera que si el usuario no está conforme con la respuesta, puede investigar el contenido citado para resolver la discrepancia.

Con esto cubrimos los aspectos más principales en lo que un sistema RAG mejora la seguridad. A continuación, vamos a ver los desafíos que estos sistemas han de afrontar.

### **Desafíos de seguridad en un RAG**

El principal riesgo de seguridad en un RAG recae sobre la fiabilidad de los LLM y sus fuentes externas de datos.

#### LLM como cajas negras

Es común la comparación entre una caja negra y un LLM, en el sentido que las fuentes de datos de un modelo LLM no siempre son fiables, debido a la falta de transparencia y de interpretación de dichos modelos al generar la respuesta, ya que los modelos más populares tienen billones y billones de parámetros, por lo que es difícil entender como el modelo llega a su respuesta.

#### Privacidad y protección de datos del usuario

Este campo surge ante el almacenamiento de datos de una empresa en nuestra *vector database*, por lo que si hay algún problema con el LLM, puede generar serios problemas de privacidad en dicha empresa. Pero sin tomar este riesgo, no podemos implementar el resto de ventajas de nuestro RAG.

No podemos evitar este problema, únicamente ser muy conscientes de ello para reducirlo lo máximo posible.

#### Alucinaciones

Una **alucinación** es una respuesta coherente por parte de un LLM, pero que es incorrecta. 

Esto es debido mayormente a la naturaleza probabilística de los LLM, debido a la probabilidad de determinar el siguiente *token* generado, y es especialmente preocupamente en contextos donde la base de conocimientos no es demasiado fuerte.

### ***Red teaming***

**Red teaming** es una metodología de *testing* en el que se simulan las potenciales amenazas que nuestra aplicación puede tener (mencionadas en el apartado anterior) con el fin de identificarlas y mitigarlas por completo. Este término surge del ámbito militar, donde se aplican estas técnicas.

El principal objetivo de un *red team* en un RAG es sobrepasar los sistemas de seguridad de una aplicación con el fin de encontrar alguna manera en que la aplicación no funcione correctamente, generando una brecha de seguridad y por lo tanto explotando y aprovechándose de dicha vulnerabilidad.

Para evaluar el rendimiento de un LLM se suelen evaluar métricas como ARC (AI Reasoning Challenge), HellaSwag y MMLU (Multitask Model Language Understanding), que se basan en el rendimiento del modelo. Sin embargo, estos *benchmarks* no abarcan medidas de seguridad, únicamente se centran en rendimiento.

### **Áreas comunes de interés para un *red team***

Estas son las estrategias que podemos implementar en un ataque *red team* para un RAG:

- **Sesgo y estereotipos**: Manipular un *chatbot* para generar respuestas sesgadas

- **Obtención de información sensible**: Obtener información personal o de la empresa a través del *chatbot*

- **Denegación de Servicio (DDoS)**: Enviar muchas peticiones o peticiones muy costosas para tirar la disponibilidad de la herramienta

- **Alucinaciones**: Debidas a un *retriever* subóptimo y que tiende siempre a estar de acuerdo con el usuario

Y entre las técnicas a implementar, destacan las siguientes:

- **Sobrepasar seguridad**:

    - **Completación de texto**

    - **Prompts sesgados**

    - ***Prompt injection (jailbraking)***: Incluir nuevas instrucciones que sobreescriben el *prompt* inicial del RAG

    - ***Gray box prompts***: Inyectar datos incorrectos en el *prompt*

    - ***Prompt probing***

- **Automatización de red team**:

    - **Lista definida manualmente**: Tener una lista de las posibles inyecciones que se pueden hacer y aumatizar su detección para su implementar su ataque

    - **Biblioteca de prompts**: Tener un almacenamiento de *prompts* que pueda automatizar las inyecciones (similar al anterior)

    - **Herramientas open source**: Esta es una opción más avanzada, como la librería de *python* conocida como *Giskard*, para hacer un escaneo de LLM, la cual está actualizada por un equipo de investigadores ML.
